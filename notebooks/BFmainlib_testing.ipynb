# Test for 5 images w/o verbose

```python
image_paths = [f'../pictures/test_input_{i}.bmp' for i in range(1, 6)]
```

```python
for i in range(5):
    fig, ax = plt.subplots(1,1, figsize=(16,12))
    test_image_class = BF_image(verbose=False)
    test_image_class.load_image(image_paths[i])
    test_image_class.preprocess_image()
    test_image_class.segment_image(type='ridges')
    ax.imshow(test_image_class.segment_draw())
    ax.axis('off')

plt.show()
```

```python
fig, axs = plt.subplots(2,3, figsize=(14,8))

for i in range(5):
    test_image_class = BF_image(verbose=False)
    test_image_class.load_image(image_paths[i])
    test_image_class.preprocess_image()
    test_image_class.segment_image(type='ridges')
    x, y, w, h = cv2.boundingRect(test_image_class.objects_db['1'].object_countour_coords)
    axs[i//3][i%3].imshow(test_image_class.object_draw(test_image_class.objects_db['1'].object_countour_coords)[y:y + h, x:x + w])
    axs[i//3][i%3].axis('off')

axs[1][2].axis('off')
plt.show()
```
